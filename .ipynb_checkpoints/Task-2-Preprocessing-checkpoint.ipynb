```python
df.columns = df.columns.str.strip().str.lower().str.replace(' ', '_').str.replace('-', '_')
print(df.columns.tolist())
```

```python
# Convert claim_date to datetime
df['claim_date'] = pd.to_datetime(df['claim_date'], infer_datetime_format=True, errors='coerce')

# Check if any failed
print(f"Failed to parse dates: {df['claim_date'].isna().sum()}")

# Extract time features (useful for fraud detection)
df['claim_month'] = df['claim_date'].dt.month
df['claim_day_of_week_encoded'] = df['claim_date'].dt.dayofweek  # Mon=0, Sun=6
df['claim_year'] = df['claim_date'].dt.year

# Drop the text version of day_of_week (redundant now) - CORRECTED COLUMN NAME
df.drop('claim_day_of_week', axis=1, inplace=True, errors='ignore')

# Verify
print(df[['claim_date', 'claim_month', 'claim_day_of_week_encoded']].head())
```

```python
print("Columns after dropping:", df.columns.tolist())
print("\nSample of new date features:")
print(df[['claim_date', 'claim_month', 'claim_day_of_week_encoded', 'claim_year']].head(10))
```

```python
# Show final column list
print(f"Total columns now: {len(df.columns)}")
```

```python
print(df['injury_claim'].head(10))

# Clean and convert
df['injury_claim'] = df['injury_claim'].astype(str).str.replace('$', '').str.replace(',', '').astype(float)
```
