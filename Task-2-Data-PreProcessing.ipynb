```python
df["claim_day"] = df["claim_date"].dt.day

df["claim_month"] = df["claim_date"].dt.month

df["claim_year"] = df["claim_date"].dt.year

df["is_weekend"] = (df["claim_date"].dt.dayofweek >= 5).astype(int)
```

```python
df["witness_present"] = df["witness_present"].fillna(
    df["witness_present"].mode()[0]
)
```

```python
numerical_columns = [
    "age_of_driver",
    "safety_rating",
    "annual_income",
    "zip_code",
    "past_num_of_claims",
    "liab_prct",
    "vehicle_price",
    "total_claim",
    "injury_claim",
    "policy deductible",
    "annual premium",
    "days open",
    "form defects",
    "claim_day",
    "claim_month",
    "claim_year"
]
```

```python
for col in numerical_columns:
    plt.figure(figsize=(5,3))
    plt.boxplot(df[col])
    plt.title(col)
    plt.show()
```

```python
for col in numerical_columns:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df[(df[col] < lower) | (df[col] > upper)]

    print(f"{col}: {len(outliers)} outliers")
```

```python
scale_columns = [
    "age_of_driver",
    "safety_rating",
    "annual_income",
    "past_num_of_claims",
    "liab_prct",
    "vehicle_price",
    "total_claim",
    "injury_claim",
    "policy deductible",
    "annual premium",
    "days open",
    "form defects"
]
```
